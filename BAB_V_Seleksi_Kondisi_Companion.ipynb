## 5.1 Concept recap: control statements
In Python, control statements are typically discussed as:

- **selection** (this chapter): making choices based on a condition  
- **iteration** (next chapter): repeating actions  
- **jumping/breaking** (often discussed with loops & functions)

Selection is the “decision engine” of a program.


## Boolean logic essentials ✅
### 1) Comparison operators
| Meaning | Operator | Example |
|---|---|---|
| equal | `==` | `x == 10` |
| not equal | `!=` | `x != 10` |
| greater/less | `>`, `<` | `x > 10` |
| greater/less or equal | `>=`, `<=` | `x <= 10` |

### 2) Logical operators
- `and`  → both must be True  
- `or`   → at least one True  
- `not`  → flips True/False


````python
# Quick sandbox: evaluate boolean expressions
exprs = [
    ("3 > 2 and 10 >= 10", 3 > 2 and 10 >= 10),
    ("(5 - 2) == 3 or False", (5 - 2) == 3 or False),
    ("not (7 < 3)", not (7 < 3)),
]
for s, v in exprs:
    print(f"{s:30} -> {v}")
````

## 5.2 Selection structures with `if` 🌱
Python “only” uses the `if` statement for selection, but it appears in several common forms.

### A) `if` (single case)
```python
if condition:
    statements
```

### B) `if-else` (two cases)
```python
if condition:
    statements_when_true
else:
    statements_when_false
```

### C) `if-elif-else` (multiple cases)
```python
if cond1:
    ...
elif cond2:
    ...
else:
    ...
```

### D) nested `if` (decisions inside decisions)
```python
if cond1:
    if cond2:
        ...
```


## Indentation is the rule 🧱
Python uses **indentation** (spaces) to define blocks.

- consistent indentation matters (commonly 4 spaces)  
- mixed tabs/spaces can break code  
- everything under `if:` that is indented belongs to that branch


````python
# Indentation demo (correct)
x = 5
if x > 3:
    print("x is bigger than 3")
    print("this line is still inside the if-block")
print("this line is outside the if-block")
````

## Pattern: interactive decision program
A very common structure:

1) read input (string)  
2) convert type (int/float)  
3) compute condition  
4) print output


````python
# Template
user_text = input("Enter a number: ")
n = int(user_text)   # try changing to float(...)
if n > 0:
    print("Positive")
elif n == 0:
    print("Zero")
else:
    print("Negative")
````

## Worked example 2: Salinity index (simple categorization) 🌊
I can map a numeric value into categories using `if-elif-else`.

Example categories (editable):
- `< 0.5`  → Fresh  
- `0.5–30` → Brackish  
- `> 30`   → Saline


````python
salt = float(input("Masukkan garam terlarut (ppt): "))

if salt < 0.5:
    idx = "Fresh"
elif salt <= 30:
    idx = "Brackish"
else:
    idx = "Saline"

print("Indeks salinitas:", idx)
````

## Nested decisions: quality label example 🧪
Sometimes a decision needs **two-stage filtering**.

Example:
- first check validity range  
- then classify into levels


````python
score = float(input("Score (0-100): "))

if 0 <= score <= 100:
    if score >= 85:
        label = "A"
    elif score >= 70:
        label = "B"
    elif score >= 55:
        label = "C"
    else:
        label = "D"
    print("Grade:", label)
else:
    print("Invalid score: must be 0..100")
````

## Common mistakes (and fixes) 🛠️
- `=` vs `==`  
  - `=` assigns, `==` compares  
- forgetting type conversion after `input()`  
  - `input()` always returns **string**  
- unreachable branches  
  - order matters in `elif` chains


````python
# Example: unreachable branch (spot the issue)
x = 90
if x >= 50:
    print("PASS")
elif x >= 80:
    print("EXCELLENT")  # never runs because x>=50 already caught it
````

### Fix
Put the stricter condition first (`x >= 80`) then the wider one (`x >= 50`).

````python
x = 90
if x >= 80:
    print("EXCELLENT")
elif x >= 50:
    print("PASS")
else:
    print("FAIL")
````

## Mini-lab: build a menu with selection 📋
Selection logic is perfect for menus.

Goal:
- show options  
- read choice  
- execute based on choice


````python
print("=== Mini Menu ===")
print("1) Add")
print("2) Subtract")
print("3) Multiply")
print("4) Divide")

choice = input("Choose (1-4): ")

a = float(input("a = "))
b = float(input("b = "))

if choice == "1":
    print("Result:", a + b)
elif choice == "2":
    print("Result:", a - b)
elif choice == "3":
    print("Result:", a * b)
elif choice == "4":
    if b == 0:
        print("Cannot divide by zero")
    else:
        print("Result:", a / b)
else:
    print("Invalid choice")
````
